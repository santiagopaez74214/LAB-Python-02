#Ejercicio 6 - Minimax

```tupla
Dado el juego conocido como NIM, implementar una solución a este usando el algoritmo minimax visto en clase

En este juego hay una pila de piedras (por ejemplo 10).

En cada turno un jugador puede quitar 1, 2 o 3 piedras.

Pierde el jugador que no puede mover (porque ya no quedan piedras).

MAX = Computadora

MIN = Humano

Dado lo anterior y el algoritmo visto en clase, modelar el juego y simularlo donde MAX sera la cpu y MIN el humado el cual interactua mediante el input, el primer turno sera el de MIN
```

````python
import math

TAMANO_INICIAL_PILAS = 10
MOVIMIENTOS_VALIDOS = [1, 2, 3]

def minimax(piedras, es_turno_max):
    if piedras == 0:
        return 1 if not es_turno_max else -1

    if es_turno_max:
        mejor_valor = -math.inf
        for movimiento in MOVIMIENTOS_VALIDOS:
            if piedras >= movimiento:
                valor = minimax(piedras - movimiento, False)
                mejor_valor = max(mejor_valor, valor)
        return mejor_valor
    else:
        mejor_valor = math.inf
        for movimiento in MOVIMIENTOS_VALIDOS:
            if piedras >= movimiento:
                valor = minimax(piedras - movimiento, True)
                mejor_valor = min(mejor_valor, valor)
        return mejor_valor

def encontrar_mejor_movimiento(piedras):
    mejor_valor = -math.inf
    mejor_movimiento = 1

    for movimiento in MOVIMIENTOS_VALIDOS:
        if piedras >= movimiento:
            valor = minimax(piedras - movimiento, False)
            if valor > mejor_valor:
                mejor_valor = valor
                mejor_movimiento = movimiento

    return mejor_movimiento

def iniciar_juego():
    piedras_actuales = TAMANO_INICIAL_PILAS
    print("===================================")
    print(" Juego de NIM (10) ")
    print("Movimientos válidos: 1, 2 o 3 piedras.")
    print("¡El jugador que no puede mover PIERDE!")
    print("===================================")

    turno_max = False

    while piedras_actuales > 0:
        print(f"\n--- Quedan {piedras_actuales} piedras ---")

        if not turno_max:
            print("Tu turno (jugador MIN)")
            movimiento = -1
            while movimiento not in MOVIMIENTOS_VALIDOS or movimiento > piedras_actuales:
                entrada = input("¿Cuántas piedras quieres quitar? (1, 2 o 3): ")
                if entrada.isdigit():
                    movimiento = int(entrada)

            piedras_actuales -= movimiento
            turno_max = True
        else:
            print(" Turno de la Computadora (MAX)")
            movimiento = encontrar_mejor_movimiento(piedras_actuales)
            print(f"La Computadora quita {movimiento} piedras.")
            piedras_actuales -= movimiento
            turno_max = False

    print("\n===================================")
    print("¡JUEGO TERMINADO!")
    if turno_max:
        print("¡Felicidades! ¡Has ganado!")
    else:
        print("¡La Computadora ha ganado!")
    print("===================================")

if __name__ == "__main__":
    iniciar_juego()
````
