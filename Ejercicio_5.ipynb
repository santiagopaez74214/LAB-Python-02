# Ejercicio 5 - Sorted

1. Realizar los siguientes ejercicios usando sorted.


```python
print("a. Ordenar una lista de palabras alfabéticamente.\n")
palabras = ["perro", "gato", "elefante", "pez"]
alfabeticamente = sorted(palabras)
print(alfabeticamente, "\n")


print("b. Ordenar una lista de tuplas por el primer elemento.\n")
lista_tuplas = [(10,3),(4,5),(2,3),(1,0)]
ordenar_tuplas = sorted(lista_tuplas, key = lambda x: x[0])
print(ordenar_tuplas, "\n")

print('c. Dada una lista de diccionarios, ordenar por precio (mayor a menor), y luego por nombre\n\n  productos = [\n      ("Camisa", 30),\n      ("Pantalón", 55),\n      ("Medias", 10),\n      ("Chaqueta", 80)\n  ]\n')
productos = [
      ("Camisa", 30),
      ("Pantalón", 55),
      ("Medias", 10),
      ("Chaqueta", 80)
  ]
ordenar_precio = sorted(productos, key = lambda x: x[1], reverse = True)
print(ordenar_precio, "\n")

print("d. Dada una lista de palabras, ordenarlas según su longitud.\n")
palabras = ["perro", "gato", "elefante", "pez"]
ordenar = sorted(palabras, key = len)
print(ordenar, "\n")

print("e. Dada una lista de números, ordenarlos de tal forma que aparezcan primero los pares y luego los impares.\n")
numeros = [1,2,3,4,5,6,7,8]
ordenarlos = sorted(numeros, key = lambda x: x % 2 != 0)
print(ordenarlos, "\n")

print("f. Convertir una palabra a un diccionario que cuente la aparición de cada letra, ordenar dicho diccionario según la frecuencia de aparición.\n")
palabra = "diccionario"
diccionario = {}
for i in palabra:
  if i not in diccionario:
    diccionario[i] = 0
  diccionario[i] += 1
frecuencia_ordenada = dict(sorted(diccionario.items(), key=lambda x: x[1]))
print(frecuencia_ordenada, "\n")

print('g. Dada una lista de tuplas, ordenar primero por el departamento del empleado y luego por la edad.\n\n    empleados = [\n      ("Maria", "Ventas", 30),\n      ("Luis", "Tecnología", 25),\n      ("Ana", "Ventas", 25),\n      ("Pedro", "Tecnología", 28)\n    ]')
empleados = [
      ("Maria", "Ventas", 30),
      ("Luis", "Tecnología", 25),
      ("Ana", "Ventas", 25),
      ("Pedro", "Tecnología", 28)
  ]
ordenar_departamento = sorted(empleados, key = lambda x: x[1])
print()
print(f"Ordenado por departamento: {ordenar_departamento}", "\n")
ordenar_edad = sorted(empleados, key = lambda x: x[2])
print(f"Ordenado por edad: {ordenar_edad}", "\n")
```
