# Ejercicio 4 - Filter

1. Realizar los siguientes ejercicios usando filter.


```python
print("a. Dada una lista de palabras, filtrar solo aquellas con más de 4 letras.\n")
palabras = ["Santiago","Ana", "Nicolas"]
mayores_cuatro = list(filter(lambda x: len(x) > 4, palabras))
print(mayores_cuatro, "\n")

print("b. Dada una lista, filtrar los elementos que son nulos (None).\n")
lista = [1,2,None,3,5,6,None]
nulos = list(filter(lambda x: x == None, lista))
print(nulos, "\n")

print("c. Dada una lista de palabras, filtrar aquellas que empiezan por una vocal.\n")
palabras = ["Carro","Moto", "Avion", "Bicicleta"]
empiezan = list(filter(lambda x: x.lower().startswith(("a","e","i","o","u")), palabras))
print(empiezan, "\n")

print("d. Dada una lista de palabras, filtrar aquellas que son palíndromos.\n")
palabras = ["Oso", "Ana", "Rojo", "Azul"]
palindromos = list(filter(lambda x: x.lower() == x.lower()[::-1], palabras))
print(palindromos, "\n")

print("e. Dada una lista de números, filtrar los que terminan en 5.\n")
numeros = [1,2,3,4,5,6,45,4595]
termina = list(filter(lambda x: str(x).endswith("5"), numeros))
print(termina, "\n")

print('f. Dada una lista de diccionarios, filtrar los productos con precio mayor a 100.\n\n  productos = [\n    {"nombre": "Teclado", "precio": 80},\n    {"nombre": "Mouse", "precio": 40},\n    {"nombre": "Monitor", "precio": 300},\n    {"nombre": "Webcam", "precio": 150}\n]')
productos = [
    {"nombre": "Teclado", "precio": 80},
    {"nombre": "Mouse", "precio": 40},
    {"nombre": "Monitor", "precio": 300},
    {"nombre": "Webcam", "precio": 150}
]
precio_mayor = list(filter(lambda x: x["precio"] > 100, productos))
print("\n", precio_mayor, "\n")
```
