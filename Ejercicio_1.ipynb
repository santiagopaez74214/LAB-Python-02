#Ejercicio 1 - Funciones lambdas

1. Definir las siguientes funciones usando lambdas.


````python
print("a.Crear una lambda que reciba un número y devuelva True si es múltiplo de 3, False en caso contrario.")
mult_3 = lambda x: x % 3 == 0
num = int(input("n1: "))
print(mult_3(num))
print("b. Crear una lambda que reciba un número y devuelva su cubo.")
cubo = lambda x: x**3
num = int(input("n2: "))
print(cubo(num))
print("c. Crear una lambda que reciba dos números y devuelva su producto.")
producto = lambda x, y: x*y
num = int(input("num1: "))
num1 = int(input("num2: "))
print(producto(num, num1))
print("d. Crear una lambda que reciba dos números y devuelva el mayor.")
mayor = lambda x, y: max(x, y)
num = int(input("num1: "))
num1 = int(input("num2: "))
print(mayor(num, num1))
print("e. Crear una lambda que reciba una palabra y devuelva True si empieza con la letra “A" "(o “a”).")
inicia_con = lambda x: x.startswith("A") or x.startswith("a")
palabra = input("Palabra: ")
print(inicia_con(palabra))
print("f. Crear una lambda que reciba una temperatura en Celsius y la convierta a Fahrenheit.")
convertir = lambda x: (x*9/5) + 32
temperatura = int(input("temperatura: "))
print(convertir(temperatura))
````

2. Realizar el siguiente ejercicio usando lambdas

Crear una lista que contenga tres lambdas:

* Una que duplique un número
* Una que le sume 10
* Una que calcule su cuadrado

Luego, pedir un número al usuario y aplicar cada lambda a ese número, mostrando los resultados.

```plain
Ingrese un número: 5
10
15
25
```


````python
num = int(input("Ingrese un número: "))
lista = [lambda x: x*2, lambda x: x + 10, lambda x: x**2]
for operacion in lista:
  print(operacion(num))
````
