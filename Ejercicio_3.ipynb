# Ejercicio 3 - Map

1. Realizar los siguientes ejercicios usando map.


```python
print("a. Dada una lista de temperaturas en Celsius, convertirlas a Fahrenheit.\n")
def conversion (a):
  c = (a*9/5) + 32
  return c
lista = [0, 5, 100, 30, 90]
convertir =list(map(conversion, lista))
print(convertir, "\n")

print("b. Dadas dos listas con la misma cantidad de elementos, generar una nueva sumando cada elemento (primer elemento de lista 1 + primer elemento lista 2).\n")
lista1 = [1,2,3,4,5]
lista2 = [6,7,8,9,10]
sumar_listas = list(map(lambda x, y: x + y, lista1, lista2))
print(sumar_listas, "\n")

print("c. Dada una lista de radios de circulos, generar la lista de sus áreas.\n")
import math
radio = [1,2,9,16]
areas = list(map(lambda x: round(x**2*math.pi,2), radio))
print(areas, "\n")

print('d. Dado un diccionario, generar una lista de tuplas (nombre, valor+10%)\n\n  {\"pan\": 1000, \"leche\": 2500, \"café\": 5000}\n')
diccionario =  {"pan": 1000, "leche": 2500, "café": 5000}
lista_tupla = [(x,diccionario[x]) for x in diccionario]
print(lista_tupla, "\n")

print("e. Dada una matriz, multiplicar todos los elementos por 10.")
lista_matriz = [[1,2,3], [4,5,6], [7,8,9]]
aplanar = [num for fila in lista_matriz for num in fila]
multiplicar_10 = list(map(lambda x: x*10, aplanar))
print(multiplicar_10, "\n")
```
