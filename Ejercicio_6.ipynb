# Ejercicio 6 sorted-Otras funciones

1. Realizar los siguientes ejercicios usando sorted.


```python
print("a. Filtrar números pares y luego elevarlos al cuadrado.\n")
numeros = [10, 3, 2, 7, 5, 4, 9, 8, 1]
filtrados = list(filter(lambda x: x % 2 == 0, numeros))
elevados = list(map(lambda x: x ** 2, filtrados))
resultado = sorted(elevados)
print(resultado, "\n")

print("b. Dada una lista de numeros, elevar cada número al cubo, filtrar los mayores que 100, convertirlos a cadenas.\n")
numeros = [2, 5, 8, 1, 10, 4, 6]
cubos = map(lambda x: x ** 3, numeros)
mayores_que_100 = filter(lambda x: x > 100, cubos)
cadenas = map(lambda x: str(x), mayores_que_100)
resultado_b = list(sorted(cadenas))
print(resultado_b, "\n")

print("c. Dada una lista de numeros, filtrar números mayores que 10, elevarlos al cuadrado, ordenarlos de mayor a menor.\n")
numeros = [15,60,3,2,1,90]
mayores_que_10 = list(filter(lambda x: x > 10, numeros))
cuadrados = list(map(lambda x: x ** 2, mayores_que_10))
resultado_c = sorted(cuadrados, reverse=True)
print(resultado_c, "\n")
```
